```python
class SimpleEnvironment:
    def __init__(self):
        self.state = "start"
        self.goal = "treasure"

    def get_observation(self):
        return f"You are at {self.state}"

    def take_action(self, action):
        if self.state == "start" and action == "go right":
            self.state = "forest"
        elif self.state == "forest" and action == "go forward":
            self.state = "treasure"
        return self.get_observation()

    def is_done(self):
        return self.state == self.goal
```

```python
class ReActAgent:
    def __init__(self):
        self.memory = []

    def reason(self, observation):
        print("\n--- ReAct Step ---")
        print(f"Observation: {observation}")

        if "start" in observation:
            thought = "I should explore. Going right might help."
            action = "go right"

        elif "forest" in observation:
            thought = "I am close to the goal. Move forward."
            action = "go forward"

        else:
            thought = "Goal reached."
            action = "stop"

        print(f"Thought: {thought}")
        print(f"Action: {action}")

        self.memory.append({
            "observation": observation,
            "thought": thought,
            "action": action
        })

        return action
```

```python
env = SimpleEnvironment()
agent = ReActAgent()

step = 0

while not env.is_done():
    step += 1
    print(f"\n===== Step {step} =====")

    observation = env.get_observation()
    action = agent.reason(observation)
    env.take_action(action)

print("\n✅ Final State Reached: Treasure Found!")
```
